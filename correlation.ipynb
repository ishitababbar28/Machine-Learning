```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

data = pd.read_csv("insurance2.csv")

co_mtx = data.corr(numeric_only=True)

print(co_mtx)

plt.figure(figsize=(10, 6))
sns.heatmap(co_mtx, cmap="YlGnBu", annot=True, fmt=".2f")

plt.title("Correlation Heatmap of Insurance Dataset")
plt.show()
```

```python
x = data['age'].values
y = data['bmi'].values


def estimate_coef(x, y):
    n = np.size(x)

    m_x = np.mean(x)
    m_y = np.mean(y)

    SS_xy = np.sum(x * y) - n * m_x * m_y
    SS_xx = np.sum(x * x) - n * m_x * m_x

    b_1 = SS_xy / SS_xx
    b_0 = m_y - b_1 * m_x

    return (b_0, b_1)
b_0, b_1 = estimate_coef(x, y)
print("Intercept (b0):", b_0)
print("Slope (b1):", b_1)`
```
